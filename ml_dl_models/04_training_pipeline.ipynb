# GeoSR-AI — SRM Training & Validation Pipeline Notebook

```python
import torch
import sys, os
sys.path.insert(0, '../GeoSR-AI')
from training.trainer import GeoSRTrainer
from datasets.paired_dataset import create_dataloaders
from models.srcnn import SRCNN

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
train_loader, val_loader, _ = create_dataloaders('../GeoSR-AI/data/processed', batch_size=4, patch_size=128, scale_factor=4)
model = SRCNN(in_channels=3, out_channels=3, scale_factor=4).to(device)

config = {
  'training': {'batch_size': 4, 'epochs': 2, 'learning_rate': 0.0001, 'weight_decay': 0.0},
  'loss': {'reconstruction_weight': 1.0, 'spectral_weight': 0.1, 'structural_weight': 0.05},
  'checkpoint': {'checkpoint_dir': '../GeoSR-AI/checkpoints/srcnn'}
}
trainer = GeoSRTrainer(model=model, config=config, device=device)
history = trainer.fit(train_loader, val_loader, epochs=2)
print('Done training demonstration!')
```
