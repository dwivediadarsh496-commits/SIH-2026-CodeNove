# GeoSR-AI — SRM Remote Sensing Metrics Notebook (PSNR, SSIM, RMSE, SAM)

```python
import numpy as np
from metrics import calculate_psnr, calculate_ssim, calculate_rmse, calculate_sam

img1 = np.random.uniform(0, 1, (128, 128, 3))
img2 = img1 + np.random.normal(0, 0.05, (128, 128, 3))
img2 = np.clip(img2, 0, 1)

print('PSNR (dB):', calculate_psnr(img1, img2))
print('SSIM:     ', calculate_ssim(img1, img2))
print('RMSE:     ', calculate_rmse(img1, img2))
print('SAM (deg):', calculate_sam(img1, img2))
```
